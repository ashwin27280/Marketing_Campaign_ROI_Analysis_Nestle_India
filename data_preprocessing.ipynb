### Primary Key Validation
Primary keys were checked for duplicate values using the `duplicated()` method. All primary key columns (`campaign_id`, `spend_id`, `customer_id`, `record_id`) contained unique values with zero duplicates. This confirms the integrity of each table and ensures reliable joins and analysis.

```python
set(channel_spend["campaign_id"]) - set(campaigns["campaign_id"])
```

```python
set(leads["campaign_id"]) - set(campaigns["campaign_id"])
```

```python
#Campaign duration should never be negative
(campaigns["end_date"] < campaigns["start_date"]).sum()
```

```python
#Channel Spend should never be negative
(channel_spend["spend_amount"] < 0).sum()
```

```python
#Clicks cannot exceed impressions
(channel_spend["clicks"] > channel_spend["impressions"]).sum()
```

```python
#Impressions should never be negative
(channel_spend["impressions"] < 0).sum()
```

```python
#Revenue cannot be negative
(leads["revenue_generated"] < 0).sum()
```

```python
#Conversions cannot exceed leads
(leads["conversions"] > leads["leads_generated"]).sum()
```

```python
#Leads generated should never be negative
(leads["leads_generated"] < 0).sum()
```

```python
#Convert date columns as they are still object/string:

campaigns["start_date"] = pd.to_datetime(campaigns["start_date"])
campaigns["end_date"] = pd.to_datetime(campaigns["end_date"])
```

```python
#Create a campaign duration column.

campaigns["campaign_duration"] = (campaigns["end_date"] - campaigns["start_date"]).dt.days
```

```python
campaigns.to_csv("campaigns_cleaned.csv", index=False)

channel_spend.to_csv("channel_spend_cleaned.csv", index=False)

leads.to_csv("leads_conversions_cleaned.csv", index=False)

customers.to_csv("customers_cleaned.csv", index=False)
```
